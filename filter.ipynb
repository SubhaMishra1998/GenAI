# Filter Function

## Filter will only retrieve those elements which satisfy some conditions

```python
# Syntax

# list(filter(function, iterable))
```

```python
# Retrieve only those values which are even using standard function 

def evn_check(l):
    return (l%2 ==0)

l1 = [2,4,5,1,5,1,5,6,8,0,12,11,17]

print(list(filter(evn_check, l1)))
```

```python
# Retrieve only those values which are even using lambda function 

print(list(filter(lambda x : x%2==0, l1)))
```

```python
#Rterieve values greater than 5

print(list(filter(lambda x : x>5, l1)))
```

```python
#Rterieve values greater than 5 and even

print(list(filter(lambda x: x%2 == 0 and x >5, l1)))
```

```python
#Rtrieve names whose age is greateer than 25 in a dictionary

employees = [{"Name" : "Subha", "Age" : 28},
{"Name" : "Sanjoy", "Age" : 25},
{"Name" : "Rahul", "Age" : 22}
]

def print_name(l):
    return l["Age"]>25

print(list((filter(print_name, employees))))
```
