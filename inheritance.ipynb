```python
class Vehicle:
    def __init__(self, company, type, year):
        self.company = company
        self.type = type
        self.year = year

    def get_info(self):
        return f"{self.year} {self.company} {self.type}"
    

class Car(Vehicle):
    def __init__(self, company, type, year, doors):
        super().__init__(company, type, year)
        self.doors = doors

    def get_info(self):
        return f"{super().get_info()} with {self.doors} doors"      
    

class alto(Car):

    def __init__(self, company, type, year, doors, model):
        super().__init__(company, type, year, doors)
        self.model = model
        self.company = "Maruti Suzuki"

    def get_info(self):
        return f"{super().get_info()} and model is {self.model}"
    
alto800 = alto("Maruti Suzuki", "Home Car", 2020, 4, "Alto 800")
print(alto800.get_info())
```
