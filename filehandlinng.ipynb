File Handling

Reading/ Writing;

1.Open the diary
2. Read content/ Write the content
3. Close the diary



```python
f = open("example.txt", 'r')
content = f.read()
print(content)
f.close()
```

```python
f = open("example.txt", 'w')

content = f.write("This is written using file handling")

f.close()
```

```python
f = open("example.txt", 'a')

content = f.write(" \n This is second  written using file handling")

f.close()
```

```python
## Here the limitatioin us we need to explicitly close the file, to overcome this we use with command

with open("example.txt", 'r') as f:
    content = f.read()
    print(content)
```

```python
with open("example.txt", "a") as f:
    f.write("We are learning file handling with with ")
```

```python
f = open("example.txt", 'r+')
content = f.read()
f.write("\n This is  r+ command testing")
f.close()
```

```python
# write content into a file line by line using writelines()
lines =["First Line\n","Second Line\n","Third Line\n"]

with open("example.txt",'w') as file:
    file.writelines(lines)
```

```python
# Writing and reading binary file
bfile = b"\x01"
with open("example.txt", "wb") as binary_file:
    binary_file.write(bfile)
```

```python
with open("example.txt", "rb") as binary_r:
    bcontent = binary_r.read()
    print(bcontent)
    
```

```python
#Copying file from one File(source) to another(destination)

with open("source.txt", "r") as s:
    content = s.read()
with open("dest.txt", "w") as d:
    d.write(content)
   
```

```python
#Reading total lines, words,characters using realines

with open("example2.txt", "r") as f:
    lines = f.readlines()
print(len(lines))

noofWords = sum(len(line.split()) for line in lines)
print(noofWords)

noofcharacters = sum(len(line) for line in lines)
print(noofcharacters)
```
