## Lambda function

This functions are one liner function

systax  lambda arguments :expression

Arguments : input values
Expression: the actual logic


```python

x = lambda x: x**2
```

```python
x(6)
```

```python
sums = lambda a, b: a + b

sums(5, 6)
```

```python
mul = lambda a,b,c,d:a*b*c*d
mul(2,3,4,5)
```
