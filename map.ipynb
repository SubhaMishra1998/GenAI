## Map functions

Map functions

```python
l1 = [4,5,6,7,8,9]

def sqr(x):
    return x**2

msp = map(sqr, l1)
print(list(msp))
```

<!-- Map with lambda -->

```python
lambda_map = map(lambda x: x**2, l1)
print(list(lambda_map))
```

```python
# Map function for multiple iterables
del map
n1 = [1,4,5,6]
n2 = [2,3,4,5]


print(list(map(lambda x,y: x+y, n1,n2)))
```

```python
print(list(map(int, ["1", "2", "3"])))
```

```python
employees = ["John", "Jane", "Doe"]
map_values = map(str.upper, employees)
print(list(map_values))
```

```python
# Retrieve the values in side a dictionary using map function:

emplees = [{"name": "John", "age": 30}, {"name": "Jane", "age": 25}, {"name": "Doe", "age": 35}]

def get_age(employee):
    return employee["age"]

print(list(map(get_age,emplees)))

```
